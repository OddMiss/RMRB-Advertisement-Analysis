2、客户结构差异

- 鹏鼎控股：消费电子绑定深度与转型困境
  - 苹果依赖症(58%营收)限制议价能力。鹏鼎控股绑定苹果（营收占比约58%），虽然为其消费电子用板带来了极高的技术壁垒和稳定的高毛利（27.15%），但也彻底锁死了其在通讯用板上的议价能力（毛利率仅19.03%，远低于深南电路的35.53%）。在巨头面前，全球第一的制造规模（391亿营收）并不能转化为定价权。
  - AI服务器客户(英伟达/微软)为新拓展，2025年贡献不足6%
- 深南电路：多巨头均衡合作
  - 通信端：华为5G基站核心供应商
  - 算力端：英伟达GB200认证供应商，产品用于CoWoS封装

4、产业定位差异
- 鹏鼎控股：全球PCB制造规模第一(391亿营收)，但处于价值链中游
- 深南电路：卡位"芯片级"解决方案，产品进入英伟达GPU供应链
- 深南电路有更上游的“PCB+FC-BGA载板”技术闭环，直接进入了英伟达CoWoS供应链；而鹏鼎虽然体量巨大，但仍处于价值链中游的制造端。技术代差决定了毛利率的上限。

结论：鹏鼎控股毛利率呈现"消费电子支撑、通讯业务承压、新兴业务波动"的特征，2025年整体毛利率21.53%。相较于深南电路，其差异核心在于：
1. 技术代差：深南电路在超高层PCB+FC-BGA载板领域建立技术壁垒，支撑35.53%的PCB毛利率
2. 转型节奏：鹏鼎控股AI相关业务占比仅5.41%(行业最低)，深南电路AI产品占PCB收入超40%
3. 生命周期：深南电路处于高成长通道(32.22%增速)，鹏鼎控股处于转型阵痛期(10.49%增速)
