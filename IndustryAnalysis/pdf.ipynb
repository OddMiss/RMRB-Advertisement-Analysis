PCB行业产业链构成分析

一、上游：原材料与设备

1. 核心材料

- 铜箔：占覆铜板成本42%，HVLP4/5超低轮廓铜箔用于高频高速场景。在 AI 等高频高速场景下，铜箔表面越平滑，信号在传输过程中的“摩擦阻力”（损耗）就越小。
- 树脂：决定板材耐热性与介电性能。
- 玻纤布：提供机械强度。
- 覆铜板（CCL）：由上述三者压制而成，占 PCB 成本达 40%。

2. 关键设备

- 电镀设备：决定PCB导通性与信号传输性能。
- 激光钻孔机：决定了厂家能否承接高密度的订单。



二、中游：制造加工

1. 工艺环节（制造精细度）

- 涵盖工程制版、内层/外层制作、层压、钻孔、电镀、阻焊、表面处理、测试与成型八大工段。
- mSAP工艺：是目前最尖端的微雕工艺。传统方法是“把不需要的铜冲洗掉”，而 mSAP 是“在需要的地方精准生长出铜线”，从而能实现 15μm（微米）的极窄线宽。线宽越窄，意味着同样大小的板子上能容纳更多的电路，产品附加值越高。

2. 产品类型（决定企业的毛利率）

- 高多层板： 传统板只有几层，而 AI 服务器要求 18 层以上。层数越多，内部对齐和信号干扰的控制难度呈几何级数上升。
- HDI 板： 智能手机瘦身、1.6T 高速光模块的核心标配。
- 封装基板： PCB 技术的最高峰，用于连接芯片和普通 PCB，属于高壁垒赛道。

三、下游：终端应用

1. 核心领域

- AI服务器：算力爆发导致 PCB 的层数、材料全面升级。
- 通信设备与汽车电子： 光模块 PCB 增速超 50%，反映了数据中心对流量传输的迫切需求；汽车电子则要求 PCB 具备极高的稳定性和耐受力。

2. 新兴赛道

- 先进封装：诸如 CoWoS 等芯片级封装技术，要求 PCB 与封装基板深度配合，驱动了产业向更上游延伸。
- 工业控制：工控 PCB 提供了稳定且高定制化的中端市场空间。

四、总结

* **上游（原材料）：** 铜箔、玻纤布、树脂 **$\rightarrow$** 组合成 **覆铜板（CCL）** 。
* **中游（制造加工）：** 鹏鼎控股、深南电路、沪电股份等PCB厂商。
* **下游（终端应用）：** 通信设备、消费电子、汽车电子、服务器。
